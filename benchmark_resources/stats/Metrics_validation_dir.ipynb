```python
import os
import json
from jsonschema import validate, ValidationError

SCHEMA_PATH = "../wgs_qc_metrics_jsonld.schema.json"
QC_DIR = "../qc_results_jsonld"   # directory that contains *.jsonld files

# Load the schema once
with open(SCHEMA_PATH, "r") as f:
    schema = json.load(f)

def validate_file(path):
    try:
        with open(path, "r") as f:
            data = json.load(f)
        validate(data, schema)
        print(f"[OK] {path}")
        return True
    except ValidationError as e:
        print(f"[FAIL] {path}\n  → {e.message}")
        return False
    except Exception as e:
        print(f"[ERROR] {path}\n  → {e}")
        return False

def main():
    files = [f for f in os.listdir(QC_DIR) if f.endswith(".jsonld")]
    if not files:
        print("No .jsonld QC files found.")
        return

    print(f"Checking {len(files)} JSON-LD files...\n")

    failures = 0
    for file in files:
        full_path = os.path.join(QC_DIR, file)
        ok = validate_file(full_path)
        if not ok:
            failures += 1
    
    print("\n=== SUMMARY ===")
    print(f"Total: {len(files)} | Passed: {len(files)-failures} | Failed: {failures}")

if __name__ == "__main__":
    main()
```
