```python
import json
from jsonschema import validate, ValidationError
schema = json.load(open("../wgs_qc_metrics_jsonld.schema.json"))
data = json.load(open("../result.test.jsonld"))

try:
    validate(data, schema)
    print("VALID JSON-LD QC FILE")
except ValidationError as e:
    print("INVALID:", e.message)
```
